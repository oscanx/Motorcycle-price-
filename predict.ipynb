```python
import pickle
from fastapi import FastAPI
import uvicorn
from pydantic import BaseModel
```

```python
with open('model.bin','rb') as f_in:
    pipeline = pickle.load(f_in)
```

```python
class Motorcycle_Features(BaseModel):
    seller_type:str
    owner:str
    km_driven:int
    ex_showroom_price:float
    age:int
    
```

```python
app =FastAPI(title = "motorcycle Price Prediction")
@app.post("/predict")
def predict_Selling_price(motorcycle: Motorcycle_Features):
    motorcycle_dicts = Motorcycle_Features.model_dump
    prediction = pipeline.predict(motorcycle_dicts)
    return {
        "predicted_selling_price": round(float(prediction), 2)
    }
    


             
```
